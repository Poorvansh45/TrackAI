# Tracks AI - Roadmap Agent

Goal:

Generate a personalized learning roadmap based on:

- Assessment Analysis
- Missing Prerequisites
- Learning Order

This roadmap defines WHAT the learner should study.

Timeline planning will be handled separately.

# 🧩 Skill Assessment Framework

## 👤 User
- **Who is the learner?**

---

## 🔍 Assessment Agent
- Identifies the learner’s current skills and knowledge.

---

## 📋 Prerequisite Agent
- Determines **what is missing** or needs to be learned before progressing.

---

## 🗺️ Roadmap Agent
- Defines **what should be learned** next.
- Outlines the sequence of skills or topics.

---

## ⏳ Timeline Planner
- Plans **when** each skill/topic should be learned.
- Creates a structured learning schedule.

---

➡️ **Flow:**  
User → Assessment Agent → Prerequisite Agent → Roadmap Agent → Timeline Planner


```python
from dotenv import load_dotenv

from pydantic import BaseModel, Field

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

print("Roadmap Agent Ready")
```

```python
class RoadmapPhase(BaseModel):

    phase_number: int

    phase_title: str

    objective: str

    topics: list[str]

    expected_outcomes: list[str]

    mini_project: str
```

```python
class RoadmapOutput(BaseModel):

    roadmap_title: str

    estimated_phases: int

    phases: list[RoadmapPhase]

    final_capstone_project: str

    roadmap_summary: str
```

```python
structured_llm = llm.with_structured_output(
    RoadmapOutput
)
```

```python
prerequisite_result = {

    "skill": "AI/ML",

    "current_level": "Beginner",

    "missing_prerequisites": [
        "Statistics",
        "Linear Algebra"
    ],

    "revision_topics": [
        "Python Functions",
        "NumPy"
    ],

    "learning_order": [
        "Statistics",
        "Linear Algebra",
        "Machine Learning Basics"
    ]
}
```

```python
prompt = f"""
You are an expert learning roadmap generation agent.

Your task:

Generate a structured learning roadmap.

Skill:
{prerequisite_result["skill"]}

Current Level:
{prerequisite_result["current_level"]}

Missing Prerequisites:
{prerequisite_result["missing_prerequisites"]}

Revision Topics:
{prerequisite_result["revision_topics"]}

Learning Order:
{prerequisite_result["learning_order"]}

Instructions:

1. Create logical learning phases.
2. Each phase must have:
   - title
   - objective
   - topics
   - mini project
3. End with one capstone project.
4. Keep roadmap beginner friendly.
5. Return structured output only.
"""
```

```python
roadmap = structured_llm.invoke(prompt)

roadmap
```
