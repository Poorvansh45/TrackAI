# Tracks AI - Timeline Planner Agent

Goal:

Convert roadmap into a personalized study schedule.

Inputs:

- Roadmap
- Daily Study Hours
- Weekly Availability
- Goal

Outputs:

- Estimated Duration
- Weekly Plan
- Study Schedule

```python
from dotenv import load_dotenv

from pydantic import BaseModel, Field

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

print("Timeline Planner Ready")
```

```python
class WeeklyPlan(BaseModel):

    week_number: int

    focus_area: str

    expected_hours: int

    milestones: list[str]

    recommended_resources: list[str]
```

```python
class TimelineOutput(BaseModel):

    total_duration_weeks: int

    total_estimated_hours: int

    weekly_schedule: list[WeeklyPlan]

    completion_target: str

    planner_summary: str
```

```python
structured_llm = llm.with_structured_output(
    TimelineOutput
)
```

```python
user_preferences = {

    "daily_hours": 2,

    "weekly_availability": 5,

    "learning_style": "Visual",

    "goal": "Internship"
}
```

```python
roadmap_data = {

    "roadmap_title":
        "AI/ML Roadmap",

    "estimated_phases": 4,

    "phases": [

        "Math Foundations",

        "Linear Algebra",

        "Machine Learning Basics",

        "Projects"
    ]
}
```

```python
prompt = f"""
You are an expert learning timeline planner.

Create a personalized study schedule.

Roadmap:

{roadmap_data}

User Preferences:

{user_preferences}

Instructions:

1. Create a realistic weekly schedule.

2. Adjust duration based on:
   - daily study hours
   - weekly availability
   - learning goal

3. Each week should include:
   - focus area
   - expected study hours
   - milestones

4. Personalize recommendations based on learning style:

   If learning_style is "Visual":
   - Recommend video tutorials
   - Diagrams
   - Visual explanations
   - Infographics

   If learning_style is "Reading":
   - Recommend books
   - Documentation
   - Blogs
   - Articles

   If learning_style is "Audio":
   - Recommend podcasts
   - Lectures
   - Audio explanations

   If learning_style is "Hands-On":
   - Focus heavily on projects
   - Coding exercises
   - Practical assignments
   - Build-first approach

5. If goal is:
   - Internship → prioritize projects + interview prep
   - Placement → DSA + interview preparation
   - Startup → product building + deployment
   - Freelancing → client projects + portfolio
   - Personal Growth → balanced learning 
   -

6. Return structured output only.
"""
```

```python
timeline = structured_llm.invoke(prompt)

timeline
```
