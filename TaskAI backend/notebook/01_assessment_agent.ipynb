# Tracks AI - Assessment Agent

Goal:
Analyze user assessment responses and identify:

- Current Level
- Strengths
- Weaknesses
- Skill Gaps
- Topics To Skip
- Recommended Starting Point

This notebook is the first building block of Tracks AI.

```python
from dotenv import load_dotenv
import os

from pydantic import BaseModel, Field
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

print("Gemini Initialized")
```

```python
class AssessmentOutput(BaseModel):

    current_level: str = Field(
        description="Current learner level"
    )

    start_from: str = Field(
        description="Recommended starting point"
    )

    strengths: list[str]

    weaknesses: list[str]

    skip_topics: list[str]

    recommended_focus: list[str]

    reasoning: str
```

```python
structured_llm = llm.with_structured_output(
    AssessmentOutput
)
```

```python
assessment_data = {
    "skill": "AI/ML",

    "answers": {
        "programming_experience":
            "Basic programming concepts",

        "python_knowledge":
            "Variables loops functions",

        "mathematics_background":
            "Algebra and equations",

        "data_handling":
            "Excel",

        "machine_learning":
            "Never heard ML concepts"
    }
}
```

```python
prompt = f"""
You are an expert learning assessment agent.

Your task:

1. Determine learner level.
2. Determine what topics can be skipped.
3. Determine missing prerequisites.
4. Determine recommended starting point.
5. Identify strengths and weaknesses.

Skill:
{assessment_data["skill"]}

Answers:
{assessment_data["answers"]}

Return assessment result.
"""
```

```python
result = structured_llm.invoke(prompt)

result
```
