# Tracks AI - Prerequisite Agent

Goal:

Determine:

- Missing prerequisites
- Revision topics
- Learning order
- Whether learner can skip fundamentals

```python
from dotenv import load_dotenv

from pydantic import BaseModel, Field

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
class PrerequisiteOutput(BaseModel):

    prerequisites_satisfied: bool

    missing_prerequisites: list[str]

    revision_topics: list[str]

    learning_order: list[str]

    explanation: str
```

```python
structured_llm = llm.with_structured_output(
    PrerequisiteOutput
)
```

Use Assessment Output

For now use dummy data:

```python
assessment_result = {
    "skill": "AI/ML",

    "current_level": "Beginner",

    "start_from": "Python Intermediate",

    "strengths": [
        "Python Basics"
    ],

    "weaknesses": [
        "Statistics",
        "Machine Learning"
    ]
}
```

```python
prompt = f"""
You are a prerequisite analysis agent.

Skill:
{assessment_result["skill"]}

Current Level:
{assessment_result["current_level"]}

Strengths:
{assessment_result["strengths"]}

Weaknesses:
{assessment_result["weaknesses"]}

Determine:

1. Missing prerequisites.
2. Topics needing revision.
3. Recommended learning order.
4. Whether prerequisites are satisfied.

Return structured output.
"""
```

```python
result = structured_llm.invoke(prompt)

result
```
